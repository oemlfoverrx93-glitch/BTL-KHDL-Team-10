```python
import pandas as pd
```

```python
baseline = pd.read_csv(
    "../model_reports/baseline_results.csv"
)
```

```python
boosting = pd.read_csv(
    "../model_reports/boosting_results.csv"
)
```

```python
baseline = baseline.rename(
    columns={
        "CV RMSE Mean": "CV RMSE"
    }
)
```

```python
baseline = baseline[[
    "Model",
    "CV RMSE"
]]

boosting = boosting[[
    "Model",
    "CV RMSE"
]]
```

```python
final_results = pd.concat(
    [baseline, boosting],
    ignore_index=True
)
```

```python
stacking = pd.read_csv(
    "../model_reports/stacking_results.csv"
)

final_results = pd.concat(
    [final_results, stacking],
    ignore_index=True
)

final_results
```

```python
final_results.to_csv(
    "../model_reports/final_model_summary.csv",
    index=False
)
```
